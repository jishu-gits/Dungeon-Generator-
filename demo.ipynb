```python
# Sliders
rooms_slider = IntSlider(value=10, min=5, max=30, step=1, description="Rooms:")
width_slider = IntSlider(value=60, min=40, max=80, step=2, description="Width:")
height_slider = IntSlider(value=30, min=20, max=40, step=2, description="Height:")
seed_slider = IntSlider(value=42, min=0, max=1000, step=1, description="Seed:")

# Stage dropdown
stage_dropdown = Dropdown(
    options=["rooms", "bfs", "prefabs", "final"],
    value="final",
    description="Stage:"
)

# Visualization mode dropdown
vis_mode_dropdown = Dropdown(
    options=["matplotlib", "ascii"],
    value="matplotlib",
    description="View:"
)

# Button
generate_button = Button(description="Generate Dungeon", button_style="success")

# Output widget
out = Output()
```

```python
def render_ascii(dungeon):
    """
    Render dungeon as ASCII text.
    """
    chars = {0: " ", 1: "#"}  # 0=empty, 1=wall (adjust if your encoding differs)
    for row in dungeon:
        print("".join(chars.get(cell, ".") for cell in row))
```

```python
def on_generate_clicked(b):
    out.clear_output()
    with out:
        # Generate dungeon
        dungeon = generate_dungeon(
            width=width_slider.value,
            height=height_slider.value,
            num_rooms=rooms_slider.value,
            seed=seed_slider.value,
            stage=stage_dropdown.value
        )

        # Visualization mode
        if vis_mode_dropdown.value == "matplotlib":
            plt.figure(figsize=(8, 6))
            plot_dungeon(dungeon, title=f"Dungeon ({stage_dropdown.value}) | Seed={seed_slider.value}")
            plt.title(f"Dungeon ({stage_dropdown.value}) | Seed={seed_slider.value}")
            plt.show()
        else:
            print(f"Dungeon ({stage_dropdown.value}) | Seed={seed_slider.value}")
            render_ascii(dungeon)

        # ✅ Correct evaluation calls
        conn = connectivity_ratio(dungeon)
        paths = path_metrics(dungeon)

        print(f"\nConnectivity Ratio: {conn:.2f}")
        print(f"Max Distance: {paths['max_distance']}")
        print(f"Avg Distance: {paths['average_distance']:.2f}")
```

```python
ui = VBox([
    HBox([rooms_slider, width_slider, height_slider]),
    HBox([seed_slider, stage_dropdown, vis_mode_dropdown, generate_button]),
    out
])

display(ui)
```
